```python
from langgraph.store.memory import InMemoryStore
```

```python
store=InMemoryStore()
```

```python
namespace=("user","u1")
```

## creating memory

```python
# to add memories
store.put(namespace,"1",{"data":"user plays football"})
store.put(namespace,"2",{"data":"user plays volleyball too"})
```

```python
# another namespace
namespace2=("user","u2")

# adding memory
store.put(namespace2,"1",{"data":"user eats pasta"})
store.put(namespace2,"2",{"data":"user eats rice"})
```

```python
# retrieving data/memory

store.get(namespace2,"2")
```

## retrieving all memories

```python
items=store.search(namespace)
for item in items:
    print(item.value)
```
